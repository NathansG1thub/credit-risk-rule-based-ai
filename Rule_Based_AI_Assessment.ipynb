```python
import pandas as pd

url = "https://raw.githubusercontent.com/NathansG1thub/credit-risk-rule-based-ai/main/german.data"
cols = [f"A{i}" for i in range(1, 21)] + ["class"]
df = pd.read_csv(url, sep=r"\s+", header=None, names=cols)

print(df.shape)
print(df["class"].value_counts())
df.head()
```

```python
COLS = ["checking", "duration", "history", "purpose", "amount", "savings",
        "employment", "instalment_rate", "personal_status_sex", "other_debtors",
        "residence_years", "property", "age", "other_plans", "housing",
        "n_credits", "job", "n_dependants", "telephone", "foreign_worker", "class"]

df.columns = COLS
df.head()
```

```python
DECODE = {
 "checking": {"A11": "below_0", "A12": "0_to_200", "A13": "above_200", "A14": "none"},
 "history": {"A30": "no_credits", "A31": "all_paid_this_bank", "A32": "paid_duly",
             "A33": "past_delay", "A34": "critical_other_credits"},
 "purpose": {"A40": "car_new", "A41": "car_used", "A42": "furniture", "A43": "radio_tv",
             "A44": "appliances", "A45": "repairs", "A46": "education", "A47": "vacation",
             "A48": "retraining", "A49": "business", "A410": "others"},
 "savings": {"A61": "below_100", "A62": "100_to_500", "A63": "500_to_1000",
             "A64": "above_1000", "A65": "none_unknown"},
 "employment": {"A71": "unemployed", "A72": "lt_1y", "A73": "1_to_4y",
                "A74": "4_to_7y", "A75": "gte_7y"},
 "personal_status_sex": {"A91": "male_divorced", "A92": "female_divorced_married",
                         "A93": "male_single", "A94": "male_married_widowed", "A95": "female_single"},
 "other_debtors": {"A101": "none", "A102": "co_applicant", "A103": "guarantor"},
 "property": {"A121": "real_estate", "A122": "life_insurance", "A123": "car_other",
              "A124": "none_unknown"},
 "other_plans": {"A141": "bank", "A142": "stores", "A143": "none"},
 "housing": {"A151": "rent", "A152": "own", "A153": "free"},
 "job": {"A171": "unskilled_nonresident", "A172": "unskilled_resident",
         "A173": "skilled", "A174": "management"},
 "telephone": {"A191": "none", "A192": "yes"},
 "foreign_worker": {"A201": "yes", "A202": "no"},
}

for col, mapping in DECODE.items():
    df[col] = df[col].map(mapping)

df.head()
```

```python
df["bad"] = (df["class"] == 2).astype(int)
df = df.drop(columns=["class"])   # 'bad' carries the same info, cleaner

# seatbelts: these crash loudly if something went wrong, instead of carrying on broken
assert df.shape == (1000, 21)
assert df.isna().sum().sum() == 0        # catches any typo in the decode map
assert df["bad"].value_counts().to_dict() == {0: 700, 1: 300}

print("All checks passed.")
df.head()
```

```python
def expert_system_decide(row):
    # R1
    if row['history'] == 'past_delay':
        return 1, 'R1'

    # R2
    if row['checking'] in ['below_0', '0_to_200', 'none'] and row['employment'] == 'unemployed':
        return 1, 'R2'

    # R3
    if row['instalment_rate'] >= 4 and row['savings'] in ['below_100', 'none_unknown']:
        return 1, 'R3'

    # R4
    if row['history'] == 'critical_other_credits' and row['employment'] == 'unemployed' and row['amount'] > 5000:
        return 1, 'R4'

    # R5
    if row['other_debtors'] == 'guarantor' and row['employment'] in ['lt_1y', '1_to_4y', '4_to_7y', 'gte_7y']:
        return 0, 'R5'

    # R6
    if row['history'] == 'critical_other_credits' and row['employment'] in ['4_to_7y', 'gte_7y'] and row['amount'] <= 5000:
        return 0, 'R6'

    # R7
    if (row['history'] in ['all_paid_this_bank', 'paid_duly'] and
        row['checking'] in ['0_to_200', 'above_200'] and
        row['employment'] in ['lt_1y', '1_to_4y', '4_to_7y', 'gte_7y'] and
        row['amount'] <= 1000):
        return 0, 'R7'

    # R8
    if (row['history'] in ['all_paid_this_bank', 'paid_duly'] and
        row['checking'] in ['0_to_200', 'above_200'] and
        row['employment'] in ['4_to_7y', 'gte_7y'] and
        row['amount'] > 1000):
        return 0, 'R8'

    # R9
    if row['history'] == 'no_credits' and row['employment'] in ['1_to_4y', '4_to_7y', 'gte_7y'] and row['amount'] <= 5000:
        return 0, 'R9'

    # DEFAULT
    return 1, 'DEFAULT'

# Apply the expert system to every applicant in the dataset
# axis=1 means "apply this row by row"
results = df.apply(expert_system_decide, axis=1)

# Split the results into two new columns
df['expert_decision'] = results.apply(lambda x: x[0])
df['rule_fired']      = results.apply(lambda x: x[1])

# Let's see how the judge actually behaved
print("=== DECISIONS ===")
print(df['expert_decision'].value_counts().rename({0: "Approve (0)", 1: "Decline (1)"}))

print("\n=== COVERAGE (Which rules did the work?) ===")
print(df['rule_fired'].value_counts())
```

```python
from sklearn.metrics import confusion_matrix

# labels=[0,1]: 0 = good, 1 = bad. Prediction 1 = decline.
tn, fp, fn, tp = confusion_matrix(df['bad'], df['expert_decision'], labels=[0, 1]).ravel()

print(f"Good customers approved (correct):  {tn}")
print(f"Good customers declined (cost 1):   {fp}")
print(f"BAD customers approved (cost 5):    {fn}")
print(f"Bad customers declined (correct):   {tp}")

total_cost = 5*fn + 1*fp
print("\nExpert system total cost:", total_cost)
print("Baseline approve-everyone: 1500 | Baseline decline-everyone: 700")
print("Beats decline-everyone?", tn > 5*fn, " (needs good approvals > 5 x bad approvals)")

print("\n=== BAD-RATE BY RULE (who leaks risky customers?) ===")
print(df.groupby('rule_fired')['bad'].mean().round(3).sort_values(ascending=False))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import confusion_matrix
import numpy as np

# Features = everything except the banned columns, the target, and the expert's own columns
banned = ["personal_status_sex", "foreign_worker"]
feature_cols = [c for c in df.columns
                if c not in banned + ["bad", "expert_decision", "rule_fired"]]

# One-hot encode: each category becomes a yes/no column
X = pd.get_dummies(df[feature_cols])
y = df["bad"]
print("Tree sees:", X.shape, "-> rows x yes/no questions")
```

```python
def cost(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    return 5*fn + 1*fp

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rows = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):
    tree = DecisionTreeClassifier(max_depth=3, class_weight={0: 1, 1: 5},
                                  random_state=42)
    tree.fit(X.iloc[train_idx], y.iloc[train_idx])          # studies 800
    tree_pred   = tree.predict(X.iloc[test_idx])            # exam on 200 unseen
    expert_pred = df["expert_decision"].iloc[test_idx].to_numpy()  # frozen judge, same exam

    y_te = y.iloc[test_idx]
    rows.append({"fold": fold,
                 "expert_cost": cost(y_te, expert_pred),
                 "tree_cost":   cost(y_te, tree_pred)})

res = pd.DataFrame(rows)
print(res)
print("\nTOTAL (all 1,000 applicants, each examined once):")
print(res[["expert_cost", "tree_cost"]].sum())
print("Baselines on the same exams: decline-everyone = 700, approve-everyone = 1500")
```
